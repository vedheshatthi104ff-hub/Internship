## 3. Success metric

*One metric you can defend. What number means 'good'?*






**Metric: Precision@50.**

**Here's my reasoning:** a content team can only realistically review a limited number of pages per cycle, not the whole ranked list. So the metric that actually matters is "of the top 50 pages my model flags, how many are genuinely worth reviewing?" — not overall accuracy, which can look great while being useless for a small reviewable list (most pages aren't declining, so a model could get high accuracy just by predicting "fine" for everything).

**Baseline for comparison, from the starter pipeline's real run:** the rule-based baseline scored 0.240 precision@50, while random forest scored 0.740 — meaning the model catches roughly 37 out of the top 50 vs. the baseline's 12. That's the number I'd defend as "good" — beating that baseline meaningfully.
